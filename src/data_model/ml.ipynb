```python
df["source_presence_str"] = df["source_presence"].apply(
    lambda x: ", ".join(x) if hasattr(x, "__iter__") and not isinstance(x, str) else str(x)
)

print(df.groupby("source_presence_str").size())

print("\nMedia type:")
print(df["media_type"].value_counts(dropna=False))

print("\nSource:")
print(df["source"].value_counts(dropna=False))

print("\nSource presence × media type:")
print(
    pd.crosstab(
        df["source_presence_str"],
        df["media_type"],
        margins=True
    )
)
```
